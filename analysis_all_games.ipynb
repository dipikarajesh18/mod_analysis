```python
import analysis_utils as au
import seaborn as sns
import pandas as pd
```

```python
games = ["stardewvalley", "skyrim", "cyberpunk2077"]
```

```python
game_df_list = {}
for game_name in games:
    df = au.load_mod_data_per_game(game_name)
    df = au.preprocess_mod_data(df)
    game_df_list[game_name] = df

df = au.combine_all_games_data(game_df_list)
```

```python
print(au.map_category_names(df))
```

```python
top_categories = au.get_top_categories(df, category_column="broadCategory")
top_downloads = au.get_top_by_downloads(df, category_column="broadCategory")
```

```python
broad_category_counts = df["broadCategory"].unique()
print(broad_category_counts)
```

```python
# new dataframe where columns as games and rows as broad categories


for game_name in games:
    total_downloads = df[df["game"] == game_name]["downloads"].sum()
    downloads_per_category = au.get_top_by_downloads(df, category_column="broadCategory", game_name=game_name)
    print(f"Total downloads for {game_name}: {total_downloads}")
    print(f"Downloads per category for {game_name}:")
    print(downloads_per_category)

    for category in broad_category_counts:
        if category not in downloads_per_category.index:
            downloads_per_category[category] = 0
    
    print(f"Downloads per category for {game_name} after filling missing categories:")
    print(downloads_per_category)
    print(len(downloads_per_category))

    # we want the zero categories to exist in the dataframe so that the heatmap shows them as zero instead of missing
    heatmap_df[game_name] = downloads_per_category/total_downloads
    


print(heatmap_df)
sns.heatmap(heatmap_df, cmap="Spectral")
```

```python
heatmap_df = pd.DataFrame()
for game_name in games:
    non_mod_tools_df = df[df["broadCategory"] != "Modding Tools"]
    non_mod_tools_df = non_mod_tools_df[non_mod_tools_df["broadCategory"] != "Game-Specific Mechanic Expansions"]
    non_mod_tools_df = non_mod_tools_df[non_mod_tools_df["broadCategory"] != "Animations and Visuals"]
    total_downloads = non_mod_tools_df[non_mod_tools_df["game"] == game_name]["downloads"].sum()
    downloads_per_category = au.get_top_by_downloads(non_mod_tools_df, category_column="broadCategory", game_name=game_name)
    print(f"Total downloads for {game_name}: {total_downloads}")
    print(f"Downloads per category for {game_name}:")
    print(downloads_per_category)
    heatmap_df[game_name] = downloads_per_category/total_downloads


sns.heatmap(heatmap_df, cmap="Spectral")
```
